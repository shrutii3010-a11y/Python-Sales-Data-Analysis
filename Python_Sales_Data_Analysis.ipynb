# step 5: Data Cleaning
Data Cleaning is the process of identifying and handling missing values and duplicate records to improve data quality.

```python
# Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
# Remove duplicate records
df.drop_duplicates(inplace=True)

# Check remaining duplicate records
print("Duplicates after cleaning:", df.duplicated().sum())

# Check updated dataset shape
print("Updated Dataset Shape:", df.shape)
```

```python
# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum())
```

```python
for col in df.columns:
    if col == 'Postal Code':
        df[col] = df[col].fillna(0)

    elif pd.api.types.is_numeric_dtype(df[col]):
        df[col] = df[col].fillna(df[col].median())

    else:
        df[col] = df[col].fillna('Unknown')

print("Missing Values After Cleaning:")
print(df.isnull().sum())

print("Total Missing Values:", df.isnull().sum().sum())
```

## Step 6: Data Processing

Data Processing involves converting data into a suitable format for analysis. It includes date conversion and checking data types.

```python
# Convert date columns into datetime format

df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')

print(df[['Order Date', 'Ship Date']].dtypes)
```

```python
# Extract month and year from Order Date

df['Month'] = df['Order Date'].dt.month
df['Year'] = df['Order Date'].dt.year

print(df[['Order Date', 'Month', 'Year']].head())
```

## Step 7: Exploratory Data Analysis (EDA)

EDA is used to explore and understand the dataset through statistical analysis and visualizations. It helps identify patterns, trends, and relationships in the data.

```python
# Basic statistical analysis

print("Total Sales:", df['Sales'].sum())
print("Total Profit:", df['Profit'].sum())
print("Total Quantity:", df['Quantity'].sum())
print("Average Sales:", df['Sales'].mean())
print("Average Profit:", df['Profit'].mean())
```

```python
# Category-wise Sales Analysis

category_sales = df[df['Category'] != 'Unknown'].groupby('Category')['Sales'].sum()

category_sales.plot(kind='bar', title='Category-wise Sales')
plt.xticks(rotation=0)
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.show()
```

```python
# Region-wise Sales Analysis

region_sales = df[df['Region'] != 'Unknown'].groupby('Region')['Sales'].sum()

region_sales.plot(kind='bar', title='Region-wise Sales')
plt.xlabel('Region')
plt.xticks(rotation=0)
plt.ylabel('Total Sales')
plt.show()
```

```python
# Sub-Category-wise Sales Analysis

subcat_sales = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(12, 6))
subcat_sales.plot(kind='bar')

plt.title("Sub-Category-wise Sales Analysis")
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Monthly Sales Analysis

monthly_sales = df.groupby('Month')['Sales'].sum()

plt.figure(figsize=(10, 5))
monthly_sales.plot(kind='bar')

plt.title("Monthly Sales Analysis")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
# Year-wise Sales Analysis

yearly_sales = df.groupby('Year')['Sales'].sum()

plt.figure(figsize=(8, 5))
yearly_sales.plot(kind='bar')

plt.title("Year-wise Sales Analysis")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
# Monthly Sales Trend

monthly_sales = df.groupby('Month')['Sales'].sum()

plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(range(1, 13))
plt.grid(True)
plt.show()
```

```python
# Category-wise Sales Distribution

category_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(7, 7))

plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Category-wise Sales Distribution")
plt.show()
```

```python
# Sales vs Profit Relationship

plt.figure(figsize=(9, 5))

plt.scatter(df['Sales'], df['Profit'], alpha=0.5)

plt.title("Sales vs Profit Analysis")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.grid(True)
plt.show()
```

```python
# Correlation Heatmap

plt.figure(figsize=(10, 6))

sns.heatmap(
    df.select_dtypes(include=np.number).drop(columns=['Postal Code'], errors='ignore').corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Sales Distribution Histogram

plt.figure(figsize=(9, 5))

plt.hist(df['Sales'], bins=30, edgecolor='black')

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

# step 8 : Profit Analysis

```python
# Category-wise Profit Analysis

category_profit = df.groupby('Category')['Profit'].sum()

plt.figure(figsize=(9, 5))

plt.plot(category_profit.index, category_profit.values, marker='o')

plt.title("Category-wise Profit Analysis")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.grid(True)

plt.show()
```

```python
# Top 10 Customers by Sales

customer_sales = df.groupby('Customer Name')['Sales'].sum()

top_customers = customer_sales.sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))

top_customers.plot(kind='barh')

plt.title("Top 10 Customers by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Customer Name")
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
# Top 10 Products by Sales

product_sales = df.groupby('Product Name')['Sales'].sum()

top_products = product_sales.sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))

top_products.plot(kind='barh')

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
# Discount vs Profit Analysis

plt.figure(figsize=(9, 5))

plt.scatter(df['Discount'], df['Profit'], alpha=0.5)

plt.title("Discount vs Profit Analysis")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.axhline(y=0, linestyle='--')
plt.grid(True)

plt.show()
```
